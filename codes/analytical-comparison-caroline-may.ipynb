# Process

`import nbformat`

Loads the official Jupyter library used to read and parse .ipynb files. Because notebooks are saved as JSON, nbformat safely unpacks them into Python objects so you can extract code and markdown cells without manually dealing with raw JSON parsing.

`from thefuzz import fuzz`

Imports the fuzzy string matching module from the thefuzz library. The fuzz function calculates how textually similar two strings are to one another, generating the percentage score (e.g., 58%) by comparing character sequences.

`import difflib`

Loads Python's built-in standard library for comparing sequences, such as lists of text lines. It is used to analyze line-by-line differences and pinpoint exact matching blocks of code between two separate files.

```python
import nbformat
from thefuzz import fuzz

import difflib
```

## Week 3

```python
# Week 3

def analyze_notebook(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        nb = nbformat.read(f, as_version=4)
    
    code_content = []
    markdown_content = []
    
    for cell in nb.cells:
        if cell.cell_type == 'code':
            code_content.append(cell.source)
        elif cell.cell_type == 'markdown':
            markdown_content.append(cell.source)
            
    return "\n".join(code_content), "\n".join(markdown_content)

code1, md1 = analyze_notebook('week3_bbo.ipynb')
code2, md2 = analyze_notebook('w3-functions-all-added-fidelity-metrics.ipynb')

print(f"Code Similarity: {fuzz.ratio(code1, code2)}%")
print(f"Markdown Similarity: {fuzz.ratio(md1, md2)}%")
```

```python
# Week 3

def get_code_blocks(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        nb = nbformat.read(f, as_version=4)
    
    # Collect all code lines across cells
    lines = []
    for cell in nb.cells:
        if cell.cell_type == 'code':
            lines.extend(cell.source.splitlines(keepends=True))
    return lines

code1_lines = get_code_blocks('week3_bbo.ipynb')
code2_lines = get_code_blocks('w3-functions-all-added-fidelity-metrics.ipynb')

# Find matching blocks
matcher = difflib.SequenceMatcher(None, code1_lines, code2_lines)
for block in matcher.get_matching_blocks():
    i1, i2, n = block
    if n > 2:  # Only show matching blocks longer than 2 lines to ignore tiny coincidences
        print(f"--- Match of {n} lines found ---")
        print("".join(code1_lines[i1:i1+n]))
        print("=" * 40)
```

```python
# Week 13

def analyze_notebook(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        nb = nbformat.read(f, as_version=4)
    
    code_content = []
    markdown_content = []
    
    for cell in nb.cells:
        if cell.cell_type == 'code':
            code_content.append(cell.source)
        elif cell.cell_type == 'markdown':
            markdown_content.append(cell.source)
            
    return "\n".join(code_content), "\n".join(markdown_content)

code1, md1 = analyze_notebook('week13_bbo.ipynb')
code2, md2 = analyze_notebook('w13-functions-all-added-fidelity-metrics.ipynb')

print(f"Code Similarity: {fuzz.ratio(code1, code2)}%")
print(f"Markdown Similarity: {fuzz.ratio(md1, md2)}%")
```
